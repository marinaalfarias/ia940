```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio
```

Existe uma relação entre a duração do sinal e a percepção de altura, dada por $\Delta t \Delta f > $ cte. Se o sinal for muito curto, haverá um espalhamento espectral, ou seja, ele deixa de ser percebido como um tom e passa a ser percebido como um impulso.

Variando o valor de "duration" e "frequency", dá para avaliar esta situação vocês mesmos.

```python
duration = 100 #in miliseconds
frequency = 1000
amplitude = .1
sample_rate = 44100

time = np.arange(0, duration/1000, 1/sample_rate)
sinewave = amplitude * np.sin(2 * np.pi * frequency * time)
audio = sinewave * np.hanning(len(sinewave))

plt.figure(figsize=(20, 6), dpi=80)
plt.plot(time*1000,audio)
plt.xlabel('Tempo (ms)')
plt.ylabel('Amplitude')

Audio(audio, rate=sample_rate)
```
